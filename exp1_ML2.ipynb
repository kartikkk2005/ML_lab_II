```python
# ============================================================
# DATA LOADING, CLEANING AND EXPLORATORY DATA ANALYSIS (EDA)
# Dataset: gender_submission.csv
# Libraries: NumPy, Pandas and Matplotlib
# ============================================================


# ------------------------------------------------------------
# 1. Import required libraries
# ------------------------------------------------------------

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


# ------------------------------------------------------------
# 2. Load the dataset
# ------------------------------------------------------------

df = pd.read_csv("gender_submission.csv")

print("Dataset loaded successfully!")
print()


# ------------------------------------------------------------
# 3. Display the first 5 records
# ------------------------------------------------------------

print("First 5 rows of the dataset:")
print(df.head())
print()


# ------------------------------------------------------------
# 4. Display the last 5 records
# ------------------------------------------------------------

print("Last 5 rows of the dataset:")
print(df.tail())
print()


# ------------------------------------------------------------
# 5. Display number of rows and columns
# ------------------------------------------------------------

print("Dataset shape:")
print(df.shape)
print()


# ------------------------------------------------------------
# 6. Display column names
# ------------------------------------------------------------

print("Column names:")
print(df.columns.tolist())
print()


# ------------------------------------------------------------
# 7. Display dataset information
# ------------------------------------------------------------

print("Dataset information:")
df.info()
print()


# ------------------------------------------------------------
# 8. Check data types
# ------------------------------------------------------------

print("Data types:")
print(df.dtypes)
print()


# ============================================================
# DATA CLEANING
# ============================================================


# ------------------------------------------------------------
# 9. Check for missing values
# ------------------------------------------------------------

print("Missing values in each column:")
print(df.isnull().sum())
print()


# ------------------------------------------------------------
# 10. Calculate percentage of missing values
# ------------------------------------------------------------

missing_percentage = (df.isnull().sum() / len(df)) * 100

print("Percentage of missing values:")
print(missing_percentage)
print()


# ------------------------------------------------------------
# 11. Handle missing values
# ------------------------------------------------------------

# Check whether there are missing values
if df.isnull().sum().sum() == 0:
    print("There are no missing values in the dataset.")
else:
    print("Missing values found. Handling them...")

    # Fill numerical columns with median
    numeric_columns = df.select_dtypes(include=np.number).columns

    for column in numeric_columns:
        df[column] = df[column].fillna(df[column].median())

print()


# ------------------------------------------------------------
# 12. Check missing values again
# ------------------------------------------------------------

print("Missing values after cleaning:")
print(df.isnull().sum())
print()


# ------------------------------------------------------------
# 13. Check for duplicate rows
# ------------------------------------------------------------

print("Number of duplicate rows:")
print(df.duplicated().sum())
print()


# ------------------------------------------------------------
# 14. Remove duplicate rows
# ------------------------------------------------------------

df = df.drop_duplicates()

print("Dataset shape after removing duplicates:")
print(df.shape)
print()


# ============================================================
# SUMMARY STATISTICS
# ============================================================


# ------------------------------------------------------------
# 15. Summary statistics
# ------------------------------------------------------------

print("Summary statistics:")
print(df.describe())
print()


# ------------------------------------------------------------
# 16. Mean
# ------------------------------------------------------------

print("Mean:")
print(df.mean(numeric_only=True))
print()


# ------------------------------------------------------------
# 17. Median
# ------------------------------------------------------------

print("Median:")
print(df.median(numeric_only=True))
print()


# ------------------------------------------------------------
# 18. Standard deviation
# ------------------------------------------------------------

print("Standard deviation:")
print(df.std(numeric_only=True))
print()


# ------------------------------------------------------------
# 19. Minimum values
# ------------------------------------------------------------

print("Minimum values:")
print(df.min(numeric_only=True))
print()


# ------------------------------------------------------------
# 20. Maximum values
# ------------------------------------------------------------

print("Maximum values:")
print(df.max(numeric_only=True))
print()


# ============================================================
# EXPLORATORY DATA ANALYSIS
# ============================================================


# ------------------------------------------------------------
# 21. Analyze survival data
# ------------------------------------------------------------

print("Survival counts:")
print(df["Survived"].value_counts())
print()


# ------------------------------------------------------------
# 22. Survival percentages
# ------------------------------------------------------------

survival_percentage = df["Survived"].value_counts(normalize=True) * 100

print("Survival percentages:")
print(survival_percentage)
print()


# ============================================================
# DISTRIBUTION PLOTS
# ============================================================


# ------------------------------------------------------------
# 23. Distribution of PassengerId
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

plt.hist(
    df["PassengerId"],
    bins=20,
    edgecolor="black"
)

plt.title("Distribution of Passenger ID")
plt.xlabel("Passenger ID")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 24. Distribution of Survival
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

plt.hist(
    df["Survived"],
    bins=2,
    edgecolor="black"
)

plt.title("Distribution of Survival")
plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()


# ============================================================
# BAR CHART
# ============================================================


# ------------------------------------------------------------
# 25. Survival count bar chart
# ------------------------------------------------------------

survival_counts = df["Survived"].value_counts().sort_index()

plt.figure(figsize=(8, 5))

plt.bar(
    ["Did Not Survive", "Survived"],
    survival_counts.values
)

plt.title("Titanic Survival Count")
plt.xlabel("Survival Status")
plt.ylabel("Number of Passengers")

plt.tight_layout()
plt.show()


# ============================================================
# BOXPLOTS
# ============================================================


# ------------------------------------------------------------
# 26. Passenger ID boxplot
# ------------------------------------------------------------

plt.figure(figsize=(8, 4))

plt.boxplot(
    df["PassengerId"],
    vert=False
)

plt.title("Passenger ID Boxplot")
plt.xlabel("Passenger ID")

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 27. Survival boxplot
# ------------------------------------------------------------

plt.figure(figsize=(8, 4))

plt.boxplot(
    df["Survived"],
    vert=False
)

plt.title("Survival Boxplot")
plt.xlabel("Survived")

plt.tight_layout()
plt.show()


# ============================================================
# CORRELATION ANALYSIS
# ============================================================


# ------------------------------------------------------------
# 28. Calculate correlation matrix
# ------------------------------------------------------------

correlation_matrix = df.corr(numeric_only=True)

print("Correlation matrix:")
print(correlation_matrix)
print()


# ------------------------------------------------------------
# 29. Correlation heatmap using Matplotlib
# ------------------------------------------------------------

plt.figure(figsize=(8, 6))

plt.imshow(
    correlation_matrix,
    cmap="coolwarm",
    interpolation="nearest"
)

plt.colorbar(label="Correlation")

plt.xticks(
    range(len(correlation_matrix.columns)),
    correlation_matrix.columns
)

plt.yticks(
    range(len(correlation_matrix.columns)),
    correlation_matrix.columns
)

plt.title("Correlation Heatmap")

plt.tight_layout()
plt.show()


# ============================================================
# FINAL DATASET
# ============================================================


# ------------------------------------------------------------
# 30. Display cleaned dataset
# ------------------------------------------------------------

print("Final cleaned dataset:")
print(df.head(10))
print()


# ------------------------------------------------------------
# 31. Final dataset information
# ------------------------------------------------------------

print("Final dataset shape:")
print(df.shape)

print()

print("Final dataset information:")
df.info()
```
